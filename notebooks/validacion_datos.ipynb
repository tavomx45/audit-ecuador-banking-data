catalogo de instituciones

```python
# Extraer nombres de las instituciones financieras
nombres = (
    indicadores.iloc[7, 3:]
    .dropna()
    .astype(str)
    .str.strip()
)

# Construir catálogo de instituciones
catalogo = pd.DataFrame({
    "columna_original": nombres.index,
    "institucion": nombres.values
})

# Clasificar instituciones y agregados
catalogo["tipo"] = catalogo["institucion"].apply(
    lambda x: "Banco" if x.startswith("BP ") else "Agregado"
)

# Visualizar catálogo
display(catalogo)

# Contar registros según clasificación
print("\nResumen:")
print(catalogo["tipo"].value_counts())
```

```python
# 1. Seleccionar únicamente bancos individuales

bancos = catalogo.loc[
    catalogo["tipo"] == "Banco",
    ["columna_original", "institucion"]
].copy()

# 2. Identificar las filas de los indicadores financieros

filas_indicadores = {
    "morosidad_comercial": 19,
    "morosidad_consumo": 20,
    "morosidad_vivienda": 21,
    "morosidad_microempresa": 22,
    "morosidad_total": 23,
    "cobertura_cartera_problematica": 30
}

# 3. Obtener las columnas correspondientes a bancos

columnas_bancos = bancos["columna_original"].tolist()

# 4. Extraer los valores de cada indicador

for variable, fila in filas_indicadores.items():

    valores = indicadores.iloc[fila, columnas_bancos]

    bancos[variable] = pd.to_numeric(
        valores,
        errors="coerce"
    ).to_numpy()

# 5. Extraer la fecha del boletín financiero

fecha = pd.to_datetime(indicadores.iloc[3, 2])

# Normalizar al primer día del mes
fecha = fecha.to_period("M").to_timestamp()

# 6. Organizar nuestro DataFrame

bancos.insert(0, "fecha", fecha)

bancos = bancos.rename(
    columns={"institucion": "banco"}
)

bancos = bancos.drop(
    columns="columna_original"
)

# 7. Reiniciar los índices

bancos = bancos.reset_index(drop=True)

# 8. Visualizar resultados

display(bancos.head(10))
```

```python
# Dimensiones del dataset
print("Dimensiones:", bancos.shape)

# Tipos de datos
print("\nTipos de datos:")
print(bancos.dtypes)

# Valores faltantes
print("\nValores faltantes:")
print(bancos.isna().sum())

# Registros duplicados por banco y fecha
print("\nDuplicados:")
print(bancos.duplicated(
    subset=["fecha", "banco"]
).sum())

# Comprobar el indicador de Banco Pichincha
print("\nIndicadores de Banco Pichincha:")

display(
    bancos.loc[
        bancos["banco"] == "BP PICHINCHA"
    ]
)
```

```python
# Dimensiones de la hoja
print("Dimensiones:", cartera.shape)

# Inspeccionar encabezados y primeras filas
display(cartera.iloc[:16, :10])

# Buscar filas relacionadas con cartera bruta e improductiva
patron = r"CARTERA BRUTA|CARTERA IMPRODUCTIVA"

etiquetas = cartera.iloc[:, :4].astype("string")

filas = etiquetas.apply(
    lambda columna: columna.str.contains(
        patron,
        case=False,
        regex=True,
        na=False
    )
).any(axis=1)

display(cartera.loc[filas].iloc[:, :10])
```

```python
# 1. Obtener los nombres de las instituciones en COMPOS CART

nombres_cartera = (
    cartera.iloc[6, 3:]
    .dropna()
    .astype(str)
    .str.strip()
)

# Conservar solamente los bancos individuales
nombres_cartera = nombres_cartera[
    nombres_cartera.str.startswith("BP ")
]

# 2. Identificar sus columnas originales

columnas = nombres_cartera.index

# 3. Extraer los saldos financieros

saldos = pd.DataFrame({
    "banco": nombres_cartera.values,

    "cartera_bruta": pd.to_numeric(
        cartera.loc[59, columnas],
        errors="coerce"
    ).to_numpy(),

    "cartera_improductiva": pd.to_numeric(
        cartera.loc[53, columnas],
        errors="coerce"
    ).to_numpy()
})

# 4. Calcular la morosidad

denominador = saldos["cartera_bruta"].where(
    saldos["cartera_bruta"] > 0
)

saldos["morosidad_calculada"] = (
    saldos["cartera_improductiva"] / denominador
) * 100

# 5. Comparar con la morosidad publicada en INDIC

validacion = bancos[
    ["banco", "morosidad_total"]
].merge(
    saldos,
    on="banco",
    how="left",
    validate="one_to_one"
)

# 6. Calcular diferencia absoluta en puntos porcentuales

validacion["diferencia_pp"] = (
    validacion["morosidad_total"]
    - validacion["morosidad_calculada"]
).abs()

# 7. Mostrar resultados

display(validacion)
```

```python
# Tolerancia en puntos porcentuales
tolerancia = 0.0001

# Comprobar registros no verificables
no_verificables = validacion[
    validacion["morosidad_calculada"].isna()
]

# Identificar diferencias superiores a la tolerancia
diferencias = validacion[
    validacion["diferencia_pp"] > tolerancia
]

print("Total de bancos:", len(validacion))

print(
    "Bancos sin cálculo verificable:",
    len(no_verificables)
)

print(
    "Bancos con diferencias relevantes:",
    len(diferencias)
)

print(
    "Diferencia máxima:",
    validacion["diferencia_pp"].max()
)

# Mostrar los mayores errores
display(
    validacion.sort_values(
        "diferencia_pp",
        ascending=False
    ).head(10)
)

# Mostrar bancos que requieren revisión
if not no_verificables.empty or not diferencias.empty:
    display(
        validacion.loc[
            validacion["morosidad_calculada"].isna()
            | (validacion["diferencia_pp"] > tolerancia)
        ]
    )
```

```python
import unicodedata


def normalizar_nombre(nombre):
    """Normaliza nombres de instituciones financieras."""

    nombre = unicodedata.normalize(
        "NFKC",
        str(nombre)
    )

    nombre = " ".join(nombre.upper().split())

    return nombre
```

```python
from pathlib import Path
import pandas as pd
import numpy as np

# Ruta del archivo
series_bp_2020 = "/home/gustavo-paredes/Documents/Developer/Yachay/morosidad-bancos-ecuador/data/raw/series_entidad/BP PICHINCHA DIC 2020 web.xlsx"

# Cargar únicamente los indicadores financieros
historico = pd.read_excel(
    series_bp_2020,
    sheet_name="INDICAD",
    header=None
)

# 1. Extraer fechas
fechas_originales = historico.iloc[6, 4:]

fechas = pd.to_datetime(
    fechas_originales,
    errors="coerce"
)

# Convertir fechas seriales de Excel si fuera necesario
es_numero = fechas_originales.map(
    lambda x: isinstance(
        x, (int, float, np.integer, np.floating)
    )
)

# Convertir fechas seriales de Excel únicamente si existen
if es_numero.any():
    fechas_seriales = pd.to_numeric(
        fechas_originales.loc[es_numero],
        errors="coerce"
    )

    fechas.loc[es_numero] = pd.to_datetime(
        fechas_seriales,
        unit="D",
        origin="1899-12-30",
        errors="coerce"
    )

# 2. Extraer los indicadores
serie_pichincha = pd.DataFrame({
    "fecha": fechas,
    "morosidad_total": pd.to_numeric(
        historico.iloc[78, 3:], errors="coerce"
    ) * 100,

    "cartera_bruta": pd.to_numeric(
        historico.iloc[80, 3:], errors="coerce"
    ),

    "cartera_improductiva": pd.to_numeric(
        historico.iloc[79, 3:], errors="coerce"
    ),

    "liquidez": pd.to_numeric(
        historico.iloc[28, 3:], errors="coerce"
    ) * 100,

    "roa": pd.to_numeric(
        historico.iloc[33, 3:], errors="coerce"
    ) * 100,

    "roe": pd.to_numeric(
        historico.iloc[35, 3:], errors="coerce"
    ) * 100,

    "cobertura": pd.to_numeric(
        historico.iloc[83, 3:], errors="coerce"
    ) * 100
})

# 3. Eliminar períodos sin cartera bruta válida
serie_pichincha = serie_pichincha.loc[
    serie_pichincha["cartera_bruta"] > 0
].copy()

# 4. Normalizar fechas
serie_pichincha["fecha"] = (
    pd.to_datetime(serie_pichincha["fecha"])
    .dt.to_period("M")
    .dt.to_timestamp()
)

# 5. Incorporar la institución
serie_pichincha.insert(
    1, "banco", "BP PICHINCHA"
)

# 6. Reiniciar índices y ordenar
serie_pichincha = (
    serie_pichincha
    .sort_values("fecha")
    .reset_index(drop=True)
)

# Visualización
display(serie_pichincha.head())
display(serie_pichincha.tail())
```
