```python
# 1. Obtener los nombres de las instituciones en COMPOS CART

nombres_cartera = (
    cartera.iloc[6, 3:]
    .dropna()
    .astype(str)
    .str.strip()
)

# Conservar solamente los bancos individuales
nombres_cartera = nombres_cartera[
    nombres_cartera.str.startswith("BP ")
]

# 2. Identificar sus columnas originales

columnas = nombres_cartera.index

# 3. Extraer los saldos financieros

saldos = pd.DataFrame({
    "banco": nombres_cartera.values,

    "cartera_bruta": pd.to_numeric(
        cartera.loc[59, columnas],
        errors="coerce"
    ).to_numpy(),

    "cartera_improductiva": pd.to_numeric(
        cartera.loc[53, columnas],
        errors="coerce"
    ).to_numpy()
})

# 4. Calcular la morosidad

denominador = saldos["cartera_bruta"].where(
    saldos["cartera_bruta"] > 0
)

saldos["morosidad_calculada"] = (
    saldos["cartera_improductiva"] / denominador
) * 100

# 5. Comparar con la morosidad publicada en INDIC

validacion = bancos[
    ["banco", "morosidad_total"]
].merge(
    saldos,
    on="banco",
    how="left",
    validate="one_to_one"
)

# 6. Calcular diferencia absoluta en puntos porcentuales

validacion["diferencia_pp"] = (
    validacion["morosidad_total"]
    - validacion["morosidad_calculada"]
).abs()

# 7. Mostrar resultados

display(validacion)
```

```python
# Tolerancia en puntos porcentuales
tolerancia = 0.0001

# Comprobar registros no verificables
no_verificables = validacion[
    validacion["morosidad_calculada"].isna()
]

# Identificar diferencias superiores a la tolerancia
diferencias = validacion[
    validacion["diferencia_pp"] > tolerancia
]

print("Total de bancos:", len(validacion))

print(
    "Bancos sin cálculo verificable:",
    len(no_verificables)
)

print(
    "Bancos con diferencias relevantes:",
    len(diferencias)
)

print(
    "Diferencia máxima:",
    validacion["diferencia_pp"].max()
)

# Mostrar los mayores errores
display(
    validacion.sort_values(
        "diferencia_pp",
        ascending=False
    ).head(10)
)

# Mostrar bancos que requieren revisión
if not no_verificables.empty or not diferencias.empty:
    display(
        validacion.loc[
            validacion["morosidad_calculada"].isna()
            | (validacion["diferencia_pp"] > tolerancia)
        ]
    )
```

```python
import unicodedata


def normalizar_nombre(nombre):
    """Normaliza nombres de instituciones financieras."""

    nombre = unicodedata.normalize(
        "NFKC",
        str(nombre)
    )

    nombre = " ".join(nombre.upper().split())

    return nombre
```

```python
def extraer_hoja(df, fila_fechas, columna_inicio, variables):
    """
    Extrae indicadores financieros históricos de una hoja Excel.

    Parámetros
    ----------
    df : pd.DataFrame
        Hoja original cargada con header=None.

    fila_fechas : int
        Fila que contiene las fechas.

    columna_inicio : int
        Primera columna con datos temporales.

    variables : dict
        Diccionario {nombre_variable: fila_original}.

    Retorna
    -------
    pd.DataFrame
        Serie temporal con fechas e indicadores.
    """

    # Extraer fechas válidas
    fechas = pd.to_datetime(
        df.iloc[fila_fechas, columna_inicio:],
        errors="coerce"
    ).dropna()

    # Crear DataFrame
    resultado = pd.DataFrame({
        "fecha": fechas.dt.to_period("M").dt.to_timestamp().to_numpy()
    })

    # Extraer cada indicador
    for nombre, fila in variables.items():

        valores = pd.to_numeric(
            df.loc[fila, fechas.index],
            errors="coerce"
        )

        resultado[nombre] = valores.to_numpy()

    return resultado
```

```python
def extraer_serie_banco(ruta_archivo):
    """
    Extrae los indicadores históricos de una institución
    financiera a partir de su boletín de series.
    """

    # 1. Abrir archivo
    excel = pd.ExcelFile(ruta_archivo)

    indicad = pd.read_excel(
        excel,
        sheet_name="INDICAD",
        header=None
    )

    cartera = pd.read_excel(
        excel,
        sheet_name="INDIC CARTERA",
        header=None
    )

    # 2. Identificar institución
    banco = str(indicad.iloc[1, 0]).strip()

    banco_cartera = str(cartera.iloc[1, 0]).strip()

    if banco != banco_cartera:
        raise ValueError(
            f"Instituciones diferentes: {banco} y {banco_cartera}"
        )

    # 3. Verificar indicadores principales
    if str(indicad.iloc[8, 0]).strip().upper() != "SOLVENCIA":
        raise ValueError("No se encontró SOLVENCIA en la fila esperada.")

    if str(cartera.iloc[92, 2]).strip().upper() != "MOROSIDAD TOTAL":
        raise ValueError("No se encontró MOROSIDAD TOTAL en la fila esperada.")

    # 4. Extraer indicadores financieros
    df_financiero = extraer_hoja(
        indicad,
        fila_fechas=6,
        columna_inicio=1,
        variables=variables_financieras
    )

    # 5. Extraer indicadores de cartera
    df_cartera = extraer_hoja(
        cartera,
        fila_fechas=4,
        columna_inicio=3,
        variables=variables_cartera
    )

    # 6. Comprobar fechas
    if set(df_financiero["fecha"]) != set(df_cartera["fecha"]):
        raise ValueError(
            f"Las fechas no coinciden para {banco}"
        )

    # 7. Integrar
    resultado = df_financiero.merge(
        df_cartera,
        on="fecha",
        how="inner",
        validate="one_to_one"
    )

    # 8. Incorporar banco
    resultado.insert(1, "banco", banco)

    # 9. Ordenar
    resultado = (
        resultado
        .sort_values("fecha")
        .reset_index(drop=True)
    )

    return resultado
```

```python
# Archivos históricos
archivos_prueba = [
    SERIES_DIR / "BP GUAYAQUIL 2026_09.xlsx",
    SERIES_DIR / "BP DINERS 2026_09.xlsx"
]

# Extraer series
series_adicionales = []

for archivo in archivos_prueba:

    df = extraer_serie_banco(archivo)

    series_adicionales.append(df)

    print(f"\nBanco: {df['banco'].iloc[0]}")
    print("Observaciones:", len(df))
    print("Inicio:", df["fecha"].min())
    print("Fin:", df["fecha"].max())

    print("\nValores faltantes:")
    print(df.isna().sum())
```

```python
# Integrar las tres instituciones
panel_prueba = pd.concat(
    [serie_bp_2026] + series_adicionales,
    ignore_index=True
)

# Ordenar
panel_prueba = (
    panel_prueba
    .sort_values(["banco", "fecha"])
    .reset_index(drop=True)
)

# Comprobar dimensiones
print("Dimensiones:", panel_prueba.shape)

# Resumen por institución
resumen = (
    panel_prueba
    .groupby("banco")
    .agg(
        inicio=("fecha", "min"),
        fin=("fecha", "max"),
        meses=("fecha", "count"),
        morosidad_promedio=("morosidad_total", "mean")
    )
)

display(resumen)

# Verificar duplicados
print(
    "\nDuplicados banco-mes:",
    panel_prueba.duplicated(
        subset=["banco", "fecha"]
    ).sum()
)
```
